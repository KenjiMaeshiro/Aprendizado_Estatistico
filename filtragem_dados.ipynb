# Importação

```python
import pandas as pd
import glob
import os
```

Importando arquivos de 21-23

```python
pasta_datasets = "Pi_3\\datasets"
arquivos_21_23 = glob.glob(os.path.join(pasta_datasets, "*.csv"))
print("arquivos:",arquivos_21_23)
dfs_21_23 = []
```

Importando arquivos de 24-25

```python
pasta_notas = "PI_3\\datasets\\24-25\\notas"
pasta_participantes = "PI_3\\datasets\\24-25\\participantes"
arquivos_notas = glob.glob(os.path.join(pasta_notas, "*.csv"))
arquivos_participantes = glob.glob(os.path.join(pasta_participantes, "*.csv"))
print("arquivos:",arquivos_notas)
print("arquivos:",arquivos_participantes)
dfs_24_25 = []
```

Listas

```python
municipios = [3548500, 3551108, 3541000]

colunas_notas = [
    "NU_SEQUENCIAL",
    "NU_NOTA_CN",
    "NU_NOTA_CH",
    "NU_NOTA_LC",
    "NU_NOTA_MT",
    "NU_NOTA_REDACAO"
]

colunas_participantes = [
    "NU_INSCRICAO",
    "NU_ANO",
    "TP_COR_RACA",
    "CO_MUNICIPIO_PROVA",
    "Q006",
    "Q007",
    "Q020",
    "Q021",
    "Q023"
]

colunas_21_23 = [
    "NU_INSCRICAO",
    "NU_ANO",
    "TP_COR_RACA",
    "CO_MUNICIPIO_PROVA",
    "NU_NOTA_CN",
    "NU_NOTA_CH",
    "NU_NOTA_LC",
    "NU_NOTA_MT",
    "NU_NOTA_REDACAO",
    "Q006",
    "Q007",
    "Q020",
    "Q021",
    "Q023"
]
```

# Filtrando arquivos de 21-23

```python
for i in arquivos_21_23:
    print(f"\nLendo arquivo: {i}")
    
    df = pd.read_csv(i, sep=';', encoding="latin1", usecols=colunas_21_23)
    df = df[df["CO_MUNICIPIO_PROVA"].isin(municipios)]
    display(df)
    dfs_21_23.append(df)
```

# Juntando os arquivos de 21 a 23

```python
dfs_21_23 = pd.concat(dfs_21_23, ignore_index=True)
dfs_21_23 = dfs_21_23.sort_values(by=['NU_ANO'])
display(dfs_21_23)
```

# Filtrando arquivos de 24 e 25

```python
for n, p in zip(arquivos_notas, arquivos_participantes):
    print(f"\nLendo arquivos {n} e {p}")

    df_notas = pd.read_csv(n, sep=';', encoding="latin1", usecols=colunas_notas)
    df_participantes = pd.read_csv(p, sep=';', encoding="latin1", usecols=colunas_participantes)


    df_notas = df_notas.sort_values(by=["NU_SEQUENCIAL"]).reset_index(drop=True)
    df_participantes = df_participantes.sort_values(by=['NU_INSCRICAO']).reset_index(drop=True)

    df_junto = pd.concat([df_participantes, df_notas], axis=1)
    display(df_junto)
    dfs_24_25.append(df_junto)
```

# Juntando os arquivos 24 e 25

```python
dfs_24_25 = pd.concat(dfs_24_25, ignore_index=True)
dfs_24_25 = dfs_24_25.drop(columns=["NU_SEQUENCIAL"])
dfs_24_25 = dfs_24_25[dfs_24_25["CO_MUNICIPIO_PROVA"].isin(municipios)]
display(dfs_24_25)
```

# Juntando todos os arquivos de 21-25

```python
df_final = pd.concat([dfs_21_23, dfs_24_25], ignore_index=True)
df_final = df_final.sort_values(by=['NU_ANO', 'CO_MUNICIPIO_PROVA'])
df_final.reset_index(drop=True)
display(df_final)
```

```python
df_final.to_csv(os.path.join(pasta_datasets, "Dados_Enem_2021-2025.csv"), index=False, sep=';', decimal=',', encoding='latin1')
```
